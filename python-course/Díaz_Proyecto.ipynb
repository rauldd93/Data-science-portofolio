# Proyecto de filtros sobre base de datos
La base de datos consiste datos de frecuencia de especies de líquenes que habitan rocas y variables ambientales asociadas a éstos. 
La base de datos tiene problemas que pueden ser resueltos mediante programación: cada roca debería tener una sola fila de datos de frecuencia de especies y de variables ambientales, sin embargo, por cómo fueron tomados los datos, existen varias filas con datos de cada roca.

Otra complicación tiene que ver con que algunas especies de líquenes tienen más de un dato por roca. Ésto se debe a que en ocasiones se encontraron muestras que parecían ser de especies diferentes resultaron ser la misma y otros errores humanos. Para estos casos los investigadores determinan que el valor de frecuencia máximo debería ser tomado como el valor más cercano al correcto.

Finalmente se reemplazan los valores NaN por ceros.

En el siguiente script se busca la manera de ordenar cada dato de frecuencia de cada especie de modo que quede una sola fila por roca.

```python
# Importación los paquetes

import pandas as pd
import numpy as np
```

```python
# Importación base de datos

data = pd.read_csv('multivariado1.csv',sep=';')
data
```

```python
# Extraemos los "títulos" de la base de datos
keys = data.keys()
keys
```

```python
# Estructura de la base de datos
data.shape # tenemos 4797 filas, cuando logremos obtener una fila por roca deberían ser alrededor de 400 filas que es igual al nro de rocas muestreadas
```

```python
# Creamos un array con sin los códigos repetidos de la columna "nro"
muestras = data['nro'].unique()
muestras
muestras.shape # ahora tenemos un solo dato por roca
```

```python
# Ordenación a partid de los códigos de rocas
data = data.sort_values(by=['nro'])
data
```

```python
# Obtención del número de filas de las filas y las columnas que pretendemos obtener al final
ly = len(muestras)
lx = len(keys)

lx, ly
```

```python
# Bucle principal para filtrar los datos vacíos y obtener una sola fila de datos por roca
new = [] # nueva data frame vacía
for y in range(ly):

    q = data[data['nro']==muestras[y]] # máscara boleana para comparar los valores de la columna nro con el array de datos únicos que habíamos creado antes

    line = []
    for x in range(lx):
        
        i = keys[x]

        q_unique = q[i].unique() # creamos otra variable con los datos únicos de la máscara boleana

        if (len(q_unique)==1):            
            line.append( q_unique[0]) # la condición: si la longitud de q_unique tiene un solo valor se agrega este valor a la primera línea
        else:
            
            mask = ~pd.isna( q_unique ) # creación de otra máscara boleana para los valores NaN (True donde no hay NaN y viceversa)
            if np.sum(mask) == 0: # segunda condición: si no hay valores en mask, vmax se establece como NaN (porque no hay valores)
                vmax = np.nan
            else:
                vmax = np.max( q_unique[mask] ) # si no se cumple la condición anterior se ejecuta este bloque
            line.append( vmax ) # se agrega el valor máximo a la lista vmax

    new.append(line)
```

```python
data_new = pd.DataFrame(new, columns = keys ) # creamos una nueva dataframe con los datos filtrados y los títulos previamente guardados
```

```python
data_new
data_new.shape # la nueva base de datos tiene el número correcto de filas y columnas
```

```python
data_new.fillna(0, inplace=True) # reemplazamos los valores NaN por 0
data_new
```

```python
data_new.to_csv('multivariado_nuevo.csv', sep=';') # exportamos la nueva base de datos
```
